```python
from transformers import pipeline
import torch
```

```python
#translator using facebook model 
translator = pipeline(task="translation",
                     model="facebook/nllb-200-distilled-600M",
                     torch_dtype=torch.bfloat16)
```

```python
text = """\
Hey, This is Krishna from Bangalore, \
working as a software engineer!
"""
```

```python
text_translated = translator(text,
                            src_lang="eng_Latn",
                            tgt_lang="fra_Latn")
```

```python
text_translated
```

```python
text_translated_to_hindi = translator(text,
                            src_lang="eng_Latn",
                            tgt_lang="hin_Deva")
```

```python
text_translated_to_hindi
```

```python
text_translated_to_telugu = translator(text,
                            src_lang="eng_Latn",
                            tgt_lang="tel_Telu")
```

```python
text_translated_to_telugu
```

```python
# summarizer to summarize large texts
summarizer = pipeline(task="summarization",
                     model="facebook/bart-large-cnn",
                     torch_dtype=torch.bfloat16)
```

```python
text="""Paris is the capital and most populous city of France, with
          an estimated population of 2,175,601 residents as of 2018,
          in an area of more than 105 square kilometres (41 square
          miles). The City of Paris is the centre and seat of
          government of the region and province of Île-de-France, or
          Paris Region, which has an estimated population of
          12,174,880, or about 18 percent of the population of France
          as of 2017."""
```

```python
summary = summarizer(text,
                    min_length=10,
                    max_length=100)
```
